# 1. Business Understanding

## 1.1 Business Context

Travel insurance providers offer customers protection against unexpected events during travel, such as medical emergencies, trip cancellations, baggage loss, and other travel-related incidents.

Understanding customer claim behaviour is important for insurers to assess risk, understand their customer portfolio, and make informed business decisions.

The business has historical travel insurance data containing customer and policy-related information, along with the claim outcome.

## 1.2 Business Problem

The insurance company wants to better understand the factors associated with customers making travel insurance claims.

By analysing historical insurance data, the company aims to identify patterns and relationships between customer characteristics, policy attributes, and claim outcomes.

The business is also interested in exploring whether machine learning can be used to predict the likelihood of a claim.

## 1.3 Project Objective

The objective of this project is to analyse historical travel insurance data, identify factors associated with claim occurrence, and develop a machine learning classification model to predict whether a customer is likely to make a claim.

The analysis will aim to provide insights into claim behaviour and explore how predictive modelling could support insurance risk analysis and business decision-making.

## 1.4 Machine Learning Problem

This project is formulated as a **binary classification problem**.

- **Target variable:** Claim Status
- **Positive class:** Yes — a claim was made
- **Negative class:** No — a claim was not made

The model will aim to estimate the probability of a claim based on relevant customer, travel, and policy characteristics available at the time of prediction.

## 1.5 Potential Business Value

The project could provide value to the insurance business by:

- Improving understanding of factors associated with claim occurrence.
- Supporting insurance risk analysis.
- Identifying patterns across different customer or policy segments.
- Supporting data-driven underwriting and portfolio management decisions.
- Exploring opportunities for improved risk assessment and pricing strategies.

These potential applications would need to be assessed against business requirements, regulatory considerations, fairness, and the suitability of the model for the intended use.

## 1.6 Success Criteria

The project will be considered successful if it:

- Provides a clear understanding of the dataset and claim behaviour.
- Identifies meaningful patterns and relationships in the data.
- Develops a reliable predictive model.
- Evaluates model performance using metrics appropriate to the business objective.
- Produces interpretable insights that can support business decision-making.

Model success will not be assessed using accuracy alone. Precision, recall, F1-score, ROC-AUC, and the confusion matrix will also be considered, depending on the business cost of false-positive and false-negative predictions.

## 1.7 Project Scope

The project will initially focus on:

1. Exploratory Data Analysis (EDA)
2. Data quality assessment
3. Data preparation
4. Feature engineering
5. Model development
6. Model evaluation
7. Interpretation of results
8. Business insights and recommendations
