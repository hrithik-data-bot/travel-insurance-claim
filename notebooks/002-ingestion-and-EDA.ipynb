## 2.1 Missing Values Summary

```python
missing_summary = pd.DataFrame({"Missing Count": data.isnull().sum(), "Missing Percentage": (data.isnull().sum() / len(data)) * 100})
missing_summary.sort_values(by="Missing Percentage", ascending=False)
```
